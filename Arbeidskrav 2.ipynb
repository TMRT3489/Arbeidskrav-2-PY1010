## Oppgave 3 - omregning

I koden omregnes vinkel fra grader til radianer

```python
import numpy as np #importerer numpy for å kunne bruke matematiske funksjoner
def grader_til_radianer (v_grad):
    v_rad = v_grad * np.pi / 180
    return v_rad # returnerer den beregnedde vinkelen i radianer
v_grad = float(input ('skriv inn gradtallet:'))
v_rad = grader_til_radianer(v_grad)
print("vinkelen i radianer er:", v_rad)
```

## Oppgave 5 - areal og omkrets

Koden beregner samlet areal og ytre omkrets av figuren i arbeidskravet

```python
# variablene a og b representerer trekantens kateter, variabelen a er også diameteren for halvsirkelen
def areal_og_omkrets(a,b):
    areal_trekant = (a * b) / 2 # beregner arealet av trekanten
    areal_halvsirkel = (np.pi * (a/2)**2)/2 # beregner arealet av halvsirkelen
    totalt_areal = areal_halvsirkel + areal_trekant # summerer arealet til figuren
    hypotenus = np.sqrt(a**2 + b**2) # beregner hypotenusen ved hjelp av pytagoras
    halvsirkel_bue = np.pi * (a / 2) # beregner lengden av halvsirkelens bue
    total_omkrets = b + hypotenus + halvsirkel_bue # beregner figurens ytre omkrets
    return totalt_areal, total_omkrets
areal, omkrets = areal_og_omkrets(4, 3)
print ("figurens totale areal er: ", areal) # ber koden skrive ut totale arealet
print ("figurens ytre omkrets er: ", omkrets) # ber koden skrive ut totale ytre omkrets
```

## Oppgave 6 - opprette graf

Koden tegner grafen til funksonen f(x) = -x² - 5 for x-verdier mellom -10 og 10

```python
# numpy brukes til å generere 200 punkter, mens matplotlip brukes til å fremstille grafen
import matplotlib.pyplot as plt # importerer biblioteket for grafer
x = np.linspace(-10, 10, 200) # lager 200 jevnt fordelte x-verdier. Numpy har blitt importert over, så matematiske funksjoner er tilgjengelig
y= -x**2 - 5 # beregner funksjonsverdiene
plt.plot(x, y, color="green", linewidth=2) # tegner grafen i grønn farge
plt.title("Graf for f(x) = -x² - 5") # gir grafen en tittel
plt.xlabel("x") # gir x aksen et navn
plt.ylabel("f(x)") # gir y aksen et navn
plt.grid(True) # setter opp et nett bak grafen
plt.show() # viser grafen / plottet
```
